```python
from matplotlib import pyplot as plt
import numpy as np
```

# a) Independent interaction

```python
agents = 500000
interactions = 100
state = 50

rand_mat = np.random.rand(agents,interactions)
# Covert to 1,-1
rand_mat_1 = np.where(rand_mat <= 0.5,-1,1)

final_states = state + np.sum(rand_mat_1, axis = 1)

bins = np.arange(start=min(final_states) - 1, stop=max(final_states) + 2, step=2)

# Plotting
plt.hist(final_states, bins=bins, density=True, alpha=0.7)
plt.title("a) PMF")
plt.show()
```

#### PMF : Binomial Distribution
#### CDF : Normal Distribution

# b) Independent interaction, different change

```python
agents = 500000
interactions = 100
state = 50

rand_mat = np.random.rand(agents,interactions)
# Covert to 10,-0.5
rand_mat_1 = np.where(rand_mat <= 0.5,-0.5,10)

final_states = state + np.sum(rand_mat_1, axis = 1)

bins = np.arange(start=min(final_states) - 1, stop=max(final_states) + 2,step = 10.5)

# Plotting
plt.hist(final_states, bins=bins, density=True)
plt.title("a) PMF")
plt.show()        
```

#### PMF : Binomial Distribution
#### CDF : Normal Distribution

# c) Independent interaction, scaled change

```python
agents = 500000
interactions = 10
state = 50

rand_mat = np.random.rand(agents,interactions)
# Covert to 0.7,1/0.7
rand_mat_1 = np.where(rand_mat <= 0.5,0.7,1/0.7)

final_states = state * np.prod(rand_mat_1, axis = 1)

# Plotting linear-binning
bins = np.arange(start=min(final_states) - 1, stop=max(final_states) + 2,step = 10.5)
plt.hist(final_states, bins=bins, density=True)
plt.title("a) PMF with linear-binning")
plt.show()

# Plotting log-binning
log_bins = np.logspace(start=np.log10(1), stop=np.log10(2000), num=50)
plt.hist(final_states, bins=log_bins, density=True, alpha=0.7)
plt.xscale('log')

plt.title("c) PMF with log-binning")

```

#### PMF : Right-skewed positive Distribution
#### CDF : Lognormal Distribution
