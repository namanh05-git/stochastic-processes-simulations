```python
import numpy as np
```

## a) Homogenous

```python
N = 10000
rate = 0.3
delta_t = 0.01
constant_rate = N*rate
N_target = N//2

half_times = [] # store half-time decay in each iteration
for i in range(1000):
    N_decays = 0 # amount of cells decayed
    total_steps = 0 # number of steps performed
    while N_decays < N_target: # compare number of cells DECAYED with the target half-population number of cells
        no_decays = np.random.poisson(constant_rate*delta_t)
        N_decays = N_decays + no_decays
        total_steps = total_steps + 1
    half_times.append(total_steps*delta_t)
    
mean_half_times = np.mean(half_times)
var_half_times = np.var(half_times)

print(f'Mean half-time: {mean_half_times:.5f} s')
print(f'Variance half-time: {var_half_times:.5f} s^2')
        
```

## b) Inhomogenous

```python
N = 10000
rate = 0.3
delta_t = 0.01
N_target = N//2

half_times = []
for i in range(1000):
    N_remain = N # amount of cells remained
    total_steps = 0 # number of steps performed
    while N_remain > N_target: # compare number of cells REMAINED with the target half-population number of cells
        decay_rate = N_remain*rate
        no_decays = np.random.poisson(decay_rate*delta_t)
        no_decays = min(no_decays,N_target)
        N_remain = N_remain - no_decays
        total_steps = total_steps + 1
    half_times.append(total_steps*delta_t)
    
mean_half_times = np.mean(half_times)
var_half_times = np.var(half_times)

print(f'Mean half-time: {mean_half_times:.5f} s')
print(f'Variance half-time: {var_half_times:.5f} s^2')
        
```
