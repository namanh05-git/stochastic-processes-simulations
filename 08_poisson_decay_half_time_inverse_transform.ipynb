#### a) 
The inverse distribution derivative is presented in a seperate pdf, but the function used here is 
$T = -\frac{\ln(1-U)}{\lambda}$
##### i) Simulation using Function

```python
N0 = 10000
lambda0 = 0.2
N = 1000
```

```python
def poisson_simulation (N0, lambda0):
    # Calculate the rate
    lambda1 = N0*lambda0
    t = 0.0
    for i in range(N0 // 2):
        # Generate random interevent time
        U = np.random.random()
        # Derived function
        dt = -np.log(1-U) / lambda1
        # Add to the total time
        t += dt
    return t
```

##### ii) Calculate mean and variance of half time

```python
half_times = np.zeros(N)

for i in range(N):
    half_times[i] = poisson_simulation(N0, lambda0)

half_mean = np.mean(half_times)
half_var = np.var(half_times)

print(f"Mean: {half_mean:.4f} s")
print(f"Variance: {half_var:.4f} s^2")
```

#### b) 
Derivation can be found in a seperate pdf
##### i) Simulation using function

```python
def poisson_simulation(N0, lambda0):    
    t = 0.0    
    for i in range(N0 // 2):
        U = np.random.random()
        # Unit-rate exponential random variable
        E = -np.log(1-U)
        # Current cumulative intensity
        lambda_current = N0 * (1 - np.exp(-lambda0 * t))
        # New cumulative intensity
        lambda_next = lambda_current + E
        # Invert Lambda(t)
        t = -(1 / lambda0) * np.log(1 - lambda_next / N0)
    
    return t
```

##### ii) Mean and variance

```python
half_times = np.zeros(N)

for i in range(N):
    half_times[i] = poisson_simulation(N0, lambda0)

half_mean = np.mean(half_times)
half_var = np.var(half_times)

print(f"Mean: {half_mean:.4f} s")
print(f"Variance: {half_var:.4f} s^2")
```
