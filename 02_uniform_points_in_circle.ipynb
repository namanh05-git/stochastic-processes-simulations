```python
from matplotlib import pyplot as plt
import numpy as np
import math
```

#### a) Rejection Method
I used 0 as the center of the circle, and radius R = 1

```python
# Parameters for coordinate matrices
coords = 2
points = 2000
```

```python
# Generate the points
point_mat = np.random.uniform(-1, 1, size=(points, coords))
max_val = 2*math.pi
cond_out = (point_mat[:,0]**2 + point_mat[:,1]**2) > 1
filtered_mat = point_mat[~cond_out, :]
x_vals = filtered_mat[:,0]
y_vals = filtered_mat[:,1]

# Plot the circle
fig,ax = plt.subplots(figsize=(5, 5))
ax.grid(True)
ax.scatter(x_vals, y_vals)
plt.show()
```

#### b) Polar coordinates
I used r = 1, theta from uniform distribution, center at 0

```python
point_mat = np.random.uniform(-1, 1, size=(points, coords))
r = point_mat[:,0]
theta = point_mat[:,1]
x_vals = r*np.cos(2*np.pi*theta)
y_vals = r*np.sin(2*np.pi*theta)

# Plot the circle
fig,ax = plt.subplots(figsize=(5, 5))
ax.grid(True)
ax.scatter(x_vals, y_vals)
plt.show()
```

**Explanation**: The graph is more dense and cluttered in the middle because more points are squished within a smaller surface area in the middle. The area grows quadratically as the radius increases. So at r = 0.5, the area around the center is $0.75*pi$, and the outer area is  $0.25*pi$, although they all have similar number of points

#### c) Scaled polar coordinates

##### (ii) Implemetation

```python
R = 1 # choose a number for the polar coordinates
point_mat = np.random.uniform(0, 1, size=(points, coords))
r = np.sqrt(point_mat[:,0])*R
theta = point_mat[:,1]
x_vals = r*np.cos(2*np.pi*theta)
y_vals = r*np.sin(2*np.pi*theta)

# Plot the circle
fig,ax = plt.subplots(figsize=(5, 5))
ax.grid(True)
ax.scatter(x_vals, y_vals)
plt.show()
# Plot
```

#### d) Compare the outcomes of b) and c). Explain the difference
*c)* has a uniform distribution across the entire circle because in *b)*, the dots are distributed evenly along the axis, although the outer section has more space overall. *c)* used a square root during transformation (change of variable method), so more points are pulled towards the outer section, compensating the quadratically expanded area
